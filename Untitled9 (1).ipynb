```python
selected_crops = ['rice', 'wheat', 'maize']
features = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']

# Prepare the data
crop_data = df[df['label'].isin(selected_crops)].groupby('label')[features].mean()

# Setup radar chart
labels = features
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
# labels += labels[:1]  # No need to repeat the first label

# Create plot
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

for crop in crop_data.index:
    values = crop_data.loc[crop].tolist()
    # values += values[:1]  # Remove this line to avoid repeating the first value
    ax.plot(angles, values, label=crop) # Now 'angles' and 'values' have the same length
    ax.fill(angles, values, alpha=0.1)

# Configure chart
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_thetagrids(np.degrees(angles), features) # Use 'angles' instead of 'angles[:-1]'
ax.set_title('Radar Chart: Comparison of Crop Features')
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
```

```python
nutrients = ['N', 'P', 'K']
grouped = df.groupby('label')[nutrients].mean()

# Plot stacked bar chart
plt.figure(figsize=(14, 7))
bottom = np.zeros(len(grouped))

for nutrient in nutrients:
    plt.bar(grouped.index, grouped[nutrient], bottom=bottom, label=nutrient)
    bottom += grouped[nutrient]

plt.xticks(rotation=90)
plt.xlabel('Crop')
plt.ylabel('Average Nutrient Levels')
plt.title('Stacked Bar Chart of N, P, K by Crop')
plt.legend(title='Nutrient')
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
import os

# Define the output directory
output_dir = 'output'
os.makedirs(output_dir, exist_ok=True)

# Plot the tree
fig = plt.figure(figsize=(24, 12))
ax = fig.add_subplot()

plot_tree(rf_model.estimators_[0],
          feature_names=X.columns,
          class_names=rf_model.classes_,
          filled=True,
          rounded=True,
          max_depth=3,
          ax=ax)

# Add manual level indicators (Y positions are approx. and may vary slightly)
# These are for max_depth=3, so we label Level 0, 1, 2, 3
level_y_positions = [1.02, 0.76, 0.48, 0.23]  # tune these if needed
for i, y in enumerate(level_y_positions):
    ax.axhline(y=y, color='red', linestyle='--', linewidth=1)
    ax.text(-0.01, y, f"Level {i}", color='red', fontsize=14, fontweight='bold', va='center')

plt.title("Random Forest Tree Visualization with Level Labels", fontsize=18)
plt.tight_layout()
plt.savefig(f"{output_dir}/random_forest_tree_with_levels.png")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import os

# Prepare data
X = df.drop(columns=['label'])
y = df['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Initialize models
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
lr_model = LogisticRegression(max_iter=1000, random_state=42) # Increased max_iter
svm_model = SVC(random_state=42)

# Train models
rf_model.fit(X_train, y_train)
lr_model.fit(X_train, y_train)
svm_model.fit(X_train, y_train)

# Make predictions
rf_pred = rf_model.predict(X_test)
lr_pred = lr_model.predict(X_test)
svm_pred = svm_model.predict(X_test)

# Evaluate models
rf_accuracy = accuracy_score(y_test, rf_pred)
lr_accuracy = accuracy_score(y_test, lr_pred)
svm_accuracy = accuracy_score(y_test, svm_pred)

# Print accuracies
print(f"Random Forest Accuracy: {rf_accuracy}")
print(f"Logistic Regression Accuracy: {lr_accuracy}")
print(f"SVM Accuracy: {svm_accuracy}")

# Visualization (for Random Forest - you can adapt for other models)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import os

# Prepare data
X = df.drop(columns=['label'])
y = df['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Initialize models
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
lr_model = LogisticRegression(max_iter=1000, random_state=42)
svm_model = SVC(random_state=42)

# Train models
rf_model.fit(X_train, y_train)
lr_model.fit(X_train, y_train)
svm_model.fit(X_train, y_train)

# Make predictions
rf_pred = rf_model.predict(X_test)
lr_pred = lr_model.predict(X_test)
svm_pred = svm_model.predict(X_test)

# Evaluate models
rf_accuracy = accuracy_score(y_test, rf_pred)
lr_accuracy = accuracy_score(y_test, lr_pred)
svm_accuracy = accuracy_score(y_test, svm_pred)

# Create comparison chart
models = ['Random Forest', 'Logistic Regression', 'SVM']
accuracies = [rf_accuracy, lr_accuracy, svm_accuracy]

plt.figure(figsize=(10, 6))
plt.bar(models, accuracies, color=['blue', 'green', 'red'])
plt.title('Model Accuracy Comparison')
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1)  # Set y-axis limits to 0-1 for accuracy

for i, acc in enumerate(accuracies):
    plt.text(i, acc + 0.01, f'{acc:.2f}', ha='center', va='bottom')

plt.show()
```
